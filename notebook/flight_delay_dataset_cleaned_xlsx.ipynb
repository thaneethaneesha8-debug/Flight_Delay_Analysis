```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

# Upload dataset
uploaded = files.upload()
file = next(iter(uploaded))

# Load data
df = pd.read_excel(file)

# Display dataset
display(df.head())

# 1. Missing Values
df.isnull().sum().plot(kind='bar', figsize=(10,4))
plt.title('Missing Values in Flight Delay Dataset')
plt.ylabel('Count')
plt.show()

# 2. Flight Delay Distribution
delay_col = next((c for c in df.columns if 'delay' in c.lower()), None)

if delay_col:
    sns.histplot(df[delay_col].dropna(), bins=20, kde=True)
    plt.title('Flight Delay Distribution')
    plt.xlabel(delay_col)
    plt.show()

# 3. Top 10 Airlines
airline_col = next((c for c in df.columns if 'airline' in c.lower()), None)

if airline_col:
    df[airline_col].value_counts().head(10).plot(
        kind='bar', figsize=(10,4)
    )
    plt.title('Top Airlines by Number of Flights')
    plt.ylabel('Number of Flights')
    plt.xticks(rotation=45)
    plt.show()

# 4. Average Delay by Airline
if airline_col and delay_col:
    df.groupby(airline_col)[delay_col].mean().sort_values(
        ascending=False
    ).head(10).plot(kind='bar', figsize=(10,4))
    plt.title('Top 10 Airlines by Average Delay')
    plt.ylabel('Average Delay')
    plt.xticks(rotation=45)
    plt.show()

# 5. Correlation Heatmap
numeric_df = df.select_dtypes(include='number')

if not numeric_df.empty:
    plt.figure(figsize=(10,6))
    sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm',
                fmt='.2f')
    plt.title('Flight Delay Correlation Heatmap')
    plt.show()

# 6. Flight Delays by Hour
hour_col = next((c for c in df.columns if 'hour' in c.lower()), None)

if hour_col and delay_col:
    df.groupby(hour_col)[delay_col].mean().plot(
        kind='line', marker='o', figsize=(10,4)
    )
    plt.title('Average Flight Delay by Hour')
    plt.xlabel('Hour')
    plt.ylabel('Average Delay')
    plt.grid(True)
    plt.show()

print("Flight delay visualization completed!")
```
